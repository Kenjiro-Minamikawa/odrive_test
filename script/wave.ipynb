```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import chirp

signal_length_s = int(8)
dt = 0.02

num_signal = int(signal_length_s / dt)
t = np.linspace(0, signal_length_s, num_signal)

signal = chirp(t, f0=0.1, f1=5, t1=signal_length_s, method='logarithmic', phi=-90)

plt.plot(t, signal)
plt.title("Chirp Signal")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.show()

np.save("chirp_signal.npy", signal)
```
